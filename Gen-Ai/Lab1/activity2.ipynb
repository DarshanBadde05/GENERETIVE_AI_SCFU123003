```python
from transformers import AutoTokenizer
```

```python
tokenizer = AutoTokenizer.from_pretrained('bert-base-uncased')
```

```python
model = AutoTokenizer.from_pretrained("bert-base-uncased")
```

```python
text = "Replace me by any text you'd like."
```

```python
text_tokenized = tokenizer.tokenize(text)
print(text_tokenized)
```

```python
encoded_input = tokenizer.encode(text)
print(encoded_input)
```

```python
Tokenization Practical
Is one word always one token?
  No. One word can be represented by one token or multiple tokens, depending on the tokenizer.
Why are some words split?
  Words are split when the tokenizer does not have the complete word in its vocabulary. It breaks the word into smaller subword tokens that it already knows.
What happens to punctuation?
  Punctuation such as ., ,, !, and ? is usually treated as separate tokens.
Why does the model need token IDs?
  The model cannot directly understand text. Token IDs convert words or subwords into numbers, which the neural network can process as input.
```
